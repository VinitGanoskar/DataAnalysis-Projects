```python
import pandas as pd
```

```python
df=pd.read_csv("Cloud_Sense_CSV.csv")
```

```python
df.describe()
```

```python
#which service waste the most  ?
service_waste=df.groupby('service_clean').agg({
    'allocated_cost':'sum',
    'waste_cost':'sum'}).reset_index()
print(service_waste)

```

```python
#Which cloud services consume the most cost?

service_cost=df.groupby('service_clean')['actual_used_cost'].sum().reset_index()
print(service_cost)

service_cost.to_csv("serviceCost_summary.csv",index=False)
```

```python
# which region is costing the most.?

region_cost=df.groupby('region_clean')['actual_used_cost'].sum().reset_index()
region_cost.to_csv("RegionCost_summary.csv",index=False)
print(region_cost)
```

```python
# calculate waste percentage.
#df['waste_percent']=(df['waste_cost']/df['allocated_cost'])*100

df1=df[['service_clean','allocated_cost','waste_cost','waste_percent']]
df1.to_csv("WastePercent_summary.csv",index=False)
```

```python
# which resource IDs are costing the most.
top_resources=df.sort_values(by='actual_used_cost',ascending=False).head(5)
top_resources.to_csv("topResources5.csv",index=False)
top_resources
```
